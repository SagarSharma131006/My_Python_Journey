```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("retail_store_sales_cleaned.csv")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
print("Missing values:", df.isnull().sum().sum())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Total sales by category
category_sales = df.groupby("Category")["Total Spent"].sum().sort_values()
category_sales.plot(kind="barh", ax=axes[0, 0], color="steelblue")
axes[0, 0].set_title("Total Sales by Category")
axes[0, 0].set_xlabel("Total Spent")

# 2. Distribution of transaction spending
axes[0, 1].hist(df["Total Spent"], bins=20, color="orange", edgecolor="black")
axes[0, 1].set_title("Distribution of Total Spent")
axes[0, 1].set_xlabel("Total Spent")
axes[0, 1].set_ylabel("Frequency")

# 3. Box plot for outlier detection
axes[1, 0].boxplot(df["Total Spent"], showmeans=True)
axes[1, 0].set_title("Spending Outlier Detection")
axes[1, 0].set_ylabel("Total Spent")

# 4. Quantity vs Total Spent
axes[1, 1].scatter(
    df["Quantity"], df["Total Spent"],
    alpha=0.3, color="teal"
)
axes[1, 1].set_title("Quantity vs Total Spent")
axes[1, 1].set_xlabel("Quantity")
axes[1, 1].set_ylabel("Total Spent")

plt.suptitle("Retail Store Sales — Visualization Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
print("Total Transactions:", len(df))
print("Total Sales:", round(df["Total Spent"].sum(), 2))
print("Average Transaction:", round(df["Total Spent"].mean(), 2))
print("Median Transaction:", round(df["Total Spent"].median(), 2))
print("Highest Transaction:", df["Total Spent"].max())
print("Lowest Transaction:", df["Total Spent"].min())

print("\nTop Category by Sales:")
print(df.groupby("Category")["Total Spent"].sum().idxmax())
```
