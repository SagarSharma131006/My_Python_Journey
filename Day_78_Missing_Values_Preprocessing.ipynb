```python
import pandas as pd
import numpy as np

# Sample dataset with missing values
data = {
    "Age": [20, 21, np.nan, 24, 25, np.nan, 28, 30],
    "Salary": [25000, np.nan, 30000, 35000, np.nan, 40000, 45000, 50000],
    "Experience": [1, 2, 2, np.nan, 4, 5, np.nan, 7]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Detect missing values
print("\nMissing Values (True/False):")
print(df.isnull())

# Count missing values in each column
print("\nMissing Values Count:")
print(df.isnull().sum())

# Missing value percentage
print("\nMissing Values Percentage:")
print((df.isnull().sum() / len(df)) * 100)

# Total missing values
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Create two copies of the original dataset
df_mean = df.copy()
df_median = df.copy()

# Mean Imputation
for col in df_mean.columns:
    df_mean[col] = df_mean[col].fillna(df_mean[col].mean())

# Median Imputation
for col in df_median.columns:
    df_median[col] = df_median[col].fillna(df_median[col].median())

print("Mean Imputation:")
print(df_mean)

print("\nMedian Imputation:")
print(df_median)

print("\nMissing Values After Mean Imputation:")
print(df_mean.isnull().sum())

print("\nMissing Values After Median Imputation:")
print(df_median.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer

# Mean Imputer
mean_imputer = SimpleImputer(strategy="mean")

df_mean_sklearn = pd.DataFrame(
    mean_imputer.fit_transform(df),
    columns=df.columns
)

# Median Imputer
median_imputer = SimpleImputer(strategy="median")

df_median_sklearn = pd.DataFrame(
    median_imputer.fit_transform(df),
    columns=df.columns
)

print("Mean Imputation using SimpleImputer:")
print(df_mean_sklearn)

print("\nMedian Imputation using SimpleImputer:")
print(df_median_sklearn)

print("\nMissing Values After Preprocessing:")
print(df_median_sklearn.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer

# Sample categorical dataset
df_cat = pd.DataFrame({
    "City": ["Delhi", "Mumbai", np.nan, "Delhi", "Mumbai", np.nan, "Delhi"],
    "Department": ["IT", "HR", "IT", np.nan, "HR", "IT", np.nan]
})

print("Original Categorical Dataset:")
print(df_cat)

# Mode Imputation (Most Frequent Value)
mode_imputer = SimpleImputer(strategy="most_frequent")

df_cat_filled = pd.DataFrame(
    mode_imputer.fit_transform(df_cat),
    columns=df_cat.columns
)

print("\nAfter Mode Imputation:")
print(df_cat_filled)

print("\nMissing Values Remaining:")
print(df_cat_filled.isnull().sum())
```
