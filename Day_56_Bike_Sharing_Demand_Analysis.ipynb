```python
import pandas as pd

train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

print("Training Dataset Shape:", train.shape)
print("Testing Dataset Shape:", test.shape)

print("\nTraining Dataset Preview:")
display(train.head())

print("\nTesting Dataset Preview:")
display(test.head())
```

```python
print("===== TRAINING DATA INFO =====")
train.info()

print("\n===== TESTING DATA INFO =====")
test.info()

print("\n===== MISSING VALUES =====")
print("Training:\n", train.isnull().sum())
print("\nTesting:\n", test.isnull().sum())

print("\n===== DUPLICATE ROWS =====")
print("Training:", train.duplicated().sum())
print("Testing:", test.duplicated().sum())
```

```python
train["datetime"] = pd.to_datetime(train["datetime"])
test["datetime"] = pd.to_datetime(test["datetime"])

print("Training datetime type:", train["datetime"].dtype)
print("Testing datetime type:", test["datetime"].dtype)

print("\nTraining date range:")
print(train["datetime"].min(), "to", train["datetime"].max())

print("\nTesting date range:")
print(test["datetime"].min(), "to", test["datetime"].max())
```

```python
for df in [train, test]:
    df["year"] = df["datetime"].dt.year
    df["month"] = df["datetime"].dt.month
    df["day"] = df["datetime"].dt.day
    df["hour"] = df["datetime"].dt.hour
    df["weekday"] = df["datetime"].dt.dayofweek

print("New Training Dataset Shape:", train.shape)
print("New Testing Dataset Shape:", test.shape)

print("\nExtracted Features:")
display(
    train[["datetime", "year", "month", "day",
           "hour", "weekday"]].head(10)
)
```

```python
print("===== RENTAL COUNT STATISTICS =====")

display(train["count"].describe())

print("\n===== CASUAL vs REGISTERED USERS =====")

display(
    train[["casual", "registered", "count"]].describe()
)

print("\n===== VERIFY TOTAL RENTALS =====")

print(
    "Does casual + registered equal count for every row?",
    (train["casual"] + train["registered"]).equals(train["count"])
)
```

```python
train.to_csv("Bike_Sharing_Train_Processed.csv", index=False)
test.to_csv("Bike_Sharing_Test_Processed.csv", index=False)

print("Processed datasets saved successfully!")

print("\nTraining:", train.shape)
print("Testing:", test.shape)
```
