#5️⃣ Boolean Filtering

##Suppose hume sirf 75 se zyada marks chahiye.

```python
marks = np.array([45, 67, 89, 72, 95, 54, 81])

high_marks = marks[marks > 75]

print(high_marks)
```

##Multiple Conditions:

```python
marks = np.array([45, 67, 89, 72, 95, 54, 81])

result = marks[(marks >= 60) & (marks <= 90)]

print(result)
```

#1️⃣5️⃣ Complete NumPy Mini Project

##Ab ek small Student Performance Analyzer banate hain.

```python
import numpy as np

# Student marks
marks = np.array([
    [78, 85, 72, 80],
    [65, 70, 68, 75],
    [90, 92, 88, 95],
    [55, 60, 58, 62],
    [82, 79, 85, 88]
])

# Student average
student_avg = np.mean(marks, axis=1)

# Subject average
subject_avg = np.mean(marks, axis=0)

# Top student
top_index = np.argmax(student_avg)

# Lowest student
lowest_index = np.argmin(student_avg)

# Passing students
passing_students = student_avg >= 60

print("===== STUDENT PERFORMANCE =====")

print("\nStudent Averages:")
print(np.round(student_avg, 2))

print("\nSubject Averages:")
print(np.round(subject_avg, 2))

print("\nTop Student Index:", top_index)
print("Top Student Average:", round(student_avg[top_index], 2))

print("\nLowest Student Index:", lowest_index)
print("Lowest Student Average:", round(student_avg[lowest_index], 2))

print("\nPassing Status:")
print(passing_students)

print("\nOverall Average:")
print(round(np.mean(marks), 2))
```
