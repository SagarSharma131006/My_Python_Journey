#1️⃣ What is a DataFrame?

##DataFrame = 2-dimensional tabular data structure

##Isko Excel table jaisa samajh sakte ho.

##Example:

Name	Age	Marks

Rahul	20	85

Aman	21	72

Priya	20	91

##Yahan:

Name, Age, Marks → columns

Rahul, Aman, Priya → records/rows

0, 1, 2 → default index

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Priya"],
    "Age": [20, 21, 20],
    "Marks": [85, 72, 91]
})

print(df)
```

#2️⃣ DataFrame from Dictionary

##Ye DataFrame banane ka common method hai.

```python
data = {
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 72, 91, 68]
}

df = pd.DataFrame(data)

print(df)
```

#3️⃣ DataFrame from List of Lists

##DataFrame nested lists se bhi bana sakte hain.

```python
data = [
    ["Rahul", 20, 85],
    ["Aman", 21, 72],
    ["Priya", 20, 91]
]

df = pd.DataFrame(
    data,
    columns=["Name", "Age", "Marks"]
)

print(df)
```

#1️⃣9️⃣ Practical Student Dataset

##Ab ek proper practical example karte hain.

```python
students = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Rohit"],
    "Age": [20, 21, 20, 22, 21],
    "Python": [85, 72, 91, 64, 78],
    "Math": [80, 68, 94, 70, 75],
    "DSA": [82, 70, 89, 65, 80]
})

print(students)
```

```python
students["Average"] = students[
    ["Python", "Math", "DSA"]
].mean(axis=1)

print(students)
```

#2️⃣0️⃣ Filter High Performers

##Suppose average ≥ 75 wale students chahiye:

```python
high_performers = students[
    students["Average"] >= 75
]

print(high_performers)
```

#2️⃣1️⃣ Highest Python Score

```python
highest_python = students["Python"].max()

print("Highest Python score:", highest_python)
```

#2️⃣2️⃣ Student with Highest Average

```python
top_student = students.loc[
    students["Average"].idxmax()
]

print(top_student)
```

```python
import pandas as pd

students = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Rohit"],
    "Age": [20, 21, 20, 22, 21],
    "Python": [85, 72, 91, 64, 78],
    "Math": [80, 68, 94, 70, 75],
    "DSA": [82, 70, 89, 65, 80]
})

# Calculate the average of the three subjects
students["Average"] = students[["Python", "Math", "DSA"]].mean(axis=1)

print("--- Average Column ---")
print(students["Average"])

# Corrected column name from "Marks" to "Average"
print("\nMax Average:", students["Average"].max())
print("Min Average:", students["Average"].min())
print("Sum of Averages:", students["Average"].sum())
```

#🔥 Q10 — Mini DataFrame Analyzer

##At least 6 students ka DataFrame banao with:

###Name
###Age
###Python
###Math
###DSA

##Then:

###Har student ka average calculate karo.
###Highest average find karo.
###Lowest average find karo.
###Average ≥ 75 students filter karo.
###Highest Python marks wala student find karo.
###Sirf Name aur Average display karo.
###DataFrame ka shape print karo.
###describe() use karo.


```python
import pandas as pd

students = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Rohit"],
    "Age": [20, 21, 20, 22, 21],
    "Python": [85, 72, 91, 64, 78],
    "Math": [80, 68, 94, 70, 75],
    "DSA": [82, 70, 89, 65, 80]
})
# Calculate the average of the three subjects
students["Average"] = students[["Python", "Math", "DSA"]].mean(axis=1)

print("--- Average Column ---")
print(students["Average"])

# Corrected column name from "Marks" to "Average"
print("\nMax Average:", students["Average"].max())
print("Min Average:", students["Average"].min())
topper=students[students["Average"]>=75]
print(topper)
print(students.loc[students["Python"].idxmax()])
print(students[["Name","Average"]])
print(students.shape)
print(students.describe())
print("Sum of Averages:", students["Average"].sum())
```
