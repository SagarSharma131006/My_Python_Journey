#Dataset & Label Encoding

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Sample dataset
df = pd.DataFrame({
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Karan", "Simran"],
    "City": ["Delhi", "Mumbai", "Jaipur", "Delhi", "Mumbai", "Jaipur"],
    "Gender": ["Male", "Male", "Female", "Female", "Male", "Female"]
})

print("Original Dataset:")
print(df)

# Label Encoding
city_encoder = LabelEncoder()
gender_encoder = LabelEncoder()

df["City_Encoded"] = city_encoder.fit_transform(df["City"])
df["Gender_Encoded"] = gender_encoder.fit_transform(df["Gender"])

print("\nAfter Label Encoding:")
print(df)

# Mapping check
print("\nCity Mapping:", dict(zip(city_encoder.classes_, city_encoder.transform(city_encoder.classes_))))
print("Gender Mapping:", dict(zip(gender_encoder.classes_, gender_encoder.transform(gender_encoder.classes_))))
```

#2. One-Hot Encoding

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Original categorical columns
categorical_data = df[["City"]]

# Create OneHotEncoder
encoder = OneHotEncoder(sparse_output=False)

# Apply encoding
encoded_array = encoder.fit_transform(categorical_data)

# Convert into DataFrame
encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out(["City"]),
    index=df.index
)

print("One-Hot Encoded Data:")
print(encoded_df)

print("\nCombined Dataset:")
combined_df = pd.concat([df, encoded_df], axis=1)
print(combined_df)
```

#3. Bonus concept — drop="first"

```python
from sklearn.preprocessing import OneHotEncoder

# Normal One-Hot Encoding
normal_encoder = OneHotEncoder(sparse_output=False)
normal_result = normal_encoder.fit_transform(df[["City"]])

# One-Hot Encoding with drop="first"
drop_encoder = OneHotEncoder(sparse_output=False, drop="first")
drop_result = drop_encoder.fit_transform(df[["City"]])

print("Normal One-Hot Encoding:")
print(pd.DataFrame(
    normal_result,
    columns=normal_encoder.get_feature_names_out(),
))

print("\nOne-Hot Encoding with drop='first':")
print(pd.DataFrame(
    drop_result,
    columns=drop_encoder.get_feature_names_out(),
))
```
