# 1. np.zeros()
### np.zeros() ek aisa NumPy array create karta hai jisme saare elements 0 hote hain.

```python
import numpy as np

arr = np.zeros(5)

print(arr)
```

## 🧠 Important

###Default mein zeros ka data type generally float hota hai.

# Agar humein integers chahiye:

```python
arr = np.zeros(5, dtype=int)

print(arr)
```

# 2D Array using np.zeros()

##Hum tuple ke through shape de sakte hain:

```python
arr = np.zeros((2, 3))

print(arr)
```

# 2. np.ones()
##np.ones() ek array create karta hai jisme saare elements 1 hote hain.

```python
arr = np.ones(5)

print(arr)
```

# Integer Version:

```python
arr = np.ones(5, dtype=int)

print(arr)
```

# 2D np.ones()

```python
arr = np.ones((3, 4))

print(arr)
```

# 3. np.full()

##np.full() ka use tab karte hain jab humein har position par same value chahiye.

```python
arr = np.full(5, 10)

print(arr)
```

# 2D Example:

```python
arr = np.full((2, 3), 7)

print(arr)
```

#4. np.arange()

##Ye function range ke according values generate karta hai.

##Python ke range() ki tarah concept hai.

```python
arr = np.arange(1, 10)

print(arr)
```

#🧠 Important

##stop value include nahi hoti.

# Steps ke Sath:

```python
arr = np.arange(0, 11, 2)

print(arr)
```

# Negative Steps:

```python
arr = np.arange(10, 0, -2)

print(arr)
```

#5. np.linspace()

##np.linspace() ka use evenly spaced numbers generate karne ke liye hota hai.

```python
arr = np.linspace(0, 10, 5)

print(arr)
```

#🧠 Important Difference

##np.arange() mein hum generally step size specify karte hain.

####np.arange(0, 10, 2)

##While np.linspace() mein hum kitni values chahiye specify karte hain.

####np.linspace(0, 10, 6)

#9. np.random.randint()

##Agar humein random integers generate karne hain, toh np.random.randint() use kar sakte hain.

#🔥 Practical Example 1 — Student Marks

##Suppose humein 5 students ke marks temporarily store karne hain:

```python
marks = np.random.randint(0, 101, 5)

print(marks)
```

## Ab Day-22 ke functions ka use krke:

```python
print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
```

#🔥 Practical Example 2 — Temperature Data

##Suppose humein 7 days ka sample temperature data create karna hai:

```python
temperature = np.random.randint(20, 41, 7)

print(temperature)
```

```python
print("Average Temperature:", np.mean(temperature))
print("Maximum Temperature:", np.max(temperature))
print("Minimum Temperature:", np.min(temperature))
```
