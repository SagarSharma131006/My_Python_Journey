# 1️⃣ Import NumPy


````python
import numpy as np
````

# 2️⃣ `np.random.rand()`

Generates random floating-point values between 0 and 1.


````python
arr = np.random.rand(5)
print(arr)
````

````python
arr = np.random.rand(3, 4)
print(arr)
````

# 🧠 Important Point

`np.random.rand(3, 4)` means 3 rows and 4 columns of random decimal values.


# 3️⃣ `np.random.randint()`

Generates random integers.

### Syntax

```python
np.random.randint(start, stop, size)
```

The `stop` value is not included.


````python
arr = np.random.randint(1, 10, 5)
print(arr)
````

````python
arr = np.random.randint(1, 101, (3, 4))
print(arr)
````

# 🧠 Important Point

For `np.random.randint(1, 101, (3, 4))`, values range from 1 to 100 and the shape is 3 × 4.


# 4️⃣ `np.random.random()`

Generates random floating-point values between 0 and 1.


````python
arr = np.random.random(5)
print(arr)
````

````python
arr = np.random.random((2, 3))
print(arr)
````

# 5️⃣ `np.random.randn()`

Generates random values from a standard normal distribution. Values can be positive or negative.


````python
arr = np.random.randn(5)
print(arr)
````

````python
arr = np.random.randn(3, 4)
print(arr)
````

# 🧠 Important Point

- `rand()` → random decimals between 0 and 1
- `random()` → random decimals between 0 and 1
- `randint()` → random integers
- `randn()` → random values from a standard normal distribution


# 6️⃣ `np.random.choice()`

Selects random values from an existing array.


````python
arr = np.array([10, 20, 30, 40, 50])
print(np.random.choice(arr))
````

````python
print(np.random.choice(arr, 3))
````

````python
subjects = np.array(["Python", "NumPy", "Pandas", "SQL"])
print(np.random.choice(subjects))
````

# 7️⃣ `np.random.seed()`

A seed makes random results reproducible.


````python
np.random.seed(42)
print(np.random.randint(1, 100, 5))
````

````python
np.random.seed(42)
print(np.random.randint(1, 100, 5))
````

# 🧠 Important Point

A fixed seed is useful when we want experiments to be reproducible. You will later see the same idea with `random_state=42` in Machine Learning.


# 8️⃣ Practical Example — Student Marks


````python
np.random.seed(42)
marks = np.random.randint(0, 101, 10)
print("Marks:", marks)
````

````python
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
````

````python
print("Marks >= 80:", marks[marks >= 80])
````

# 9️⃣ Practical Example — Employee Salary


````python
np.random.seed(42)
salary = np.random.randint(20000, 100000, 10)
print("Salaries:", salary)
````

````python
print("Average Salary:", np.mean(salary))
print("Highest Salary:", np.max(salary))
print("Lowest Salary:", np.min(salary))
````

````python
print("Salaries above ₹50,000:", salary[salary > 50000])
````

# 🔟 Practical Example — Temperature


````python
np.random.seed(42)
temperature = np.random.randint(15, 41, 10)
print("Temperature:", temperature)
````

````python
print("Above 30°C:", temperature[temperature > 30])
print("Below 20°C:", temperature[temperature < 20])
````

# 1️⃣1️⃣ Generate a Random Dataset

Create a dataset with 5 students and 3 subjects.


````python
np.random.seed(42)
data = np.random.randint(1, 101, (5, 3))
print(data)
````

````python
print("Subject-wise average:")
print(np.mean(data, axis=0))
````

````python
print("Student-wise average:")
print(np.mean(data, axis=1))
````

```text
Statistics

Boolean Filtering
```
