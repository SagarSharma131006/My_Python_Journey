#2️⃣ np.sqrt() — Square Root

##np.sqrt() array ke har element ka square root calculate karta hai.

```python
arr = np.array([1, 4, 9, 16, 25])

print(np.sqrt(arr))
```

#3️⃣ np.abs() — Absolute Value

##Absolute value negative sign hata deti hai.

```python
arr = np.array([-10, -5, 0, 5, 10])

print(np.abs(arr))
```

#4️⃣ np.power() — Power Calculate Karna

##Har element ko kisi power par raise karne ke liye:

```python
arr = np.array([2, 3, 4, 5])

print(np.power(arr, 2))
```

##Cube:

```python
print(np.power(arr, 3))
```

#5️⃣ np.square()

##Square calculate karne ka direct function:

```python
arr = np.array([2, 4, 6, 8])

print(np.square(arr))
```

#6️⃣ np.round()

##Decimal values ko round karne ke liye:

```python
arr = np.array([1.2345, 2.6789, 3.4567])

print(np.round(arr))
```

#7️⃣ np.floor()

##Value ko neeche wale integer ki taraf round karta hai.

```python
arr = np.array([1.2, 2.7, 3.1, 4.9])

print(np.floor(arr))
```

#8️⃣ np.ceil()

##Value ko upar wale integer ki taraf round karta hai.

#1️⃣1️⃣ np.clip()

##np.clip() values ko ek specific range ke andar limit karta hai.

```python
arr = np.array([10, 25, 50, 75, 100, 125])

result = np.clip(arr, 20, 100)

print(result)
```

#🧠 Important Point
##np.clip(arr, minimum, maximum)

#Values:

###minimum se chhoti → minimum
###range ke andar → same
###maximum se badi → maximum

#1️⃣2️⃣ Vectorization 🔥

##Ye Day 29 ka important concept hai.

##Normally Python mein hum loop use kar sakte hain:

```python
arr = np.array([10, 20, 30, 40])

result = []

for value in arr:
    result.append(value * 2)

print(result)
```

##Lekin NumPy me Directly:

```python
arr = np.array([10, 20, 30, 40])

print(arr * 2)
```

#🧠 Vectorization kya hai?

##Simple language mein:

###Vectorization means complete array par operation ek saath perform karna instead of manually processing each element using a loop.

####arr + 10
####arr - 5
####arr * 2
####arr / 10

#1️⃣3️⃣ Vectorization Example

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr + 5)
print(arr * 2)
print(arr / 10)
```

#1️⃣8️⃣ Practical Example — Distance

#1️⃣9️⃣ Combining Multiple Functions 🔥

##NumPy functions ko chain/combine bhi kar sakte hain.

```python
data = np.array([
    -10.5,
    20.7,
    -30.2,
    40.8,
    50.4
])

result = np.abs(data)
result = np.round(result, 1)
result = np.clip(result, 0, 40)

print(result)
```

#Q9. Student Marks Analysis

```python
import numpy as np

marks = np.array([78.456, 65.782, 91.349, 88.927, 72.514])

rounded_marks = np.round(marks, 2)
highest = np.max(marks)
lowest = np.min(marks)
average = np.mean(marks)

print("Rounded Marks:", rounded_marks)
print("Highest:", highest)
print("Lowest:", lowest)
print("Average:", average)
```

#Q10. Mini Data Processing

```python
import numpy as np

arr = np.array([-15.678, 23.456, -8.234, 105.789, 67.891])

# Step 1: Absolute values
result = np.abs(arr)

# Step 2: Round
result = np.round(result, 2)

# Step 3: Clip between 0 and 100
result = np.clip(result, 0, 100)

print("Original:", arr)
print("Processed:", result)
```
