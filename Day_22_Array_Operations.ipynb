# 1. Arithmatic Operations

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(arr + 5)
print(arr - 5)
print(arr * 2)
print(arr / 2)
```

# 2. Array + Array

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

# 3. Comparison Operator

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr > 25)
print(arr < 30)
print(arr == 30)
print(arr >= 40)
```

# 4. Sum

```python
arr = np.array([10, 20, 30, 40])

print(np.sum(arr))
```

# 5. min() and max()

```python
arr = np.array([10, 50, 20, 80, 30])

print(np.min(arr))
print(np.max(arr))
```

# 6. mean()

```python
arr = np.array([10, 20, 30, 40, 50])

print(np.mean(arr))
```

# 7. std()

```python
arr = np.array([10, 20, 30, 40, 50])

print(np.std(arr))
```

# 8. axis()

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))
```

# axis=0 → columns ke according

# axis=1 → rows ke according

# Q4 — Statistics

## Given:

## marks = np.array([78, 85, 92, 67, 88, 95])

# Find:

### Total
### Minimum
### Maximum
### Average
### Standard Deviation

```python
marks=np.array([78,85,92,67,88,95])
print(marks.sum())
print(marks.min())
print(marks.max())
print(marks.mean())
print(marks.std())
```

# Q5 — Axis 🔥

## Given:

marks = np.array([
    
    [80, 70, 90],
    
    [60, 85, 75],
    
    [95, 88, 92]
])

# Find:

###Total of the complete array
###Total of each column
###Total of each row
###Average of each row

```python
marks = np.array([[80, 70, 90],[60, 85, 75],[95, 88, 92]])
print(marks.sum())
print(marks.sum(axis=0))
print(marks.sum(axis=1))
print(marks.mean(axis=1))
```
