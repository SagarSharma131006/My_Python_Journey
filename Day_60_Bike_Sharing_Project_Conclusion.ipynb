```python
import pandas as pd

train = pd.read_csv("Bike_Sharing_Day59_Dashboard_Data.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", train.shape)

print("\nColumn Names:")
print(train.columns.tolist())

print("\nFirst 5 Records:")
display(train.head())
```

```python
# 1. Monthly demand
monthly = train.groupby("month")["count"].mean()
print("HIGHEST DEMAND MONTH:", monthly.idxmax())
print("LOWEST DEMAND MONTH:", monthly.idxmin())

# 2. Hourly demand
hourly = train.groupby("hour")["count"].mean()
print("\nPEAK DEMAND HOUR:", hourly.idxmax())
print("LOWEST DEMAND HOUR:", hourly.idxmin())

# 3. Seasonal demand
seasonal = train.groupby("season_name")["count"].mean()
print("\nHIGHEST DEMAND SEASON:", seasonal.idxmax())
print("LOWEST DEMAND SEASON:", seasonal.idxmin())

# 4. Weather demand
weather = train.groupby("weather_name").agg(
    average_rentals=("count", "mean"),
    observations=("count", "size")
)

print("\nWEATHER-WISE DEMAND:")
display(weather)

# 5. User contribution
total = train["count"].sum()
casual = train["casual"].sum()
registered = train["registered"].sum()

print("\nREGISTERED USER CONTRIBUTION:",
      round(registered / total * 100, 2), "%")

print("CASUAL USER CONTRIBUTION:",
      round(casual / total * 100, 2), "%")

# 6. Working vs non-working days
working = train.groupby("workingday")["count"].mean()

print("\nWORKING DAY AVERAGE:", round(working.loc[1], 2))
print("NON-WORKING DAY AVERAGE:", round(working.loc[0], 2))
```
