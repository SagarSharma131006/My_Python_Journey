```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("retail_store_sales_cleaned.csv")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nTotal Spent Summary:")
print(df["Total Spent"].describe())
```

```python

plt.figure(figsize=(9, 5))

plt.hist(df["Total Spent"], bins=10, edgecolor="black")

plt.title("Distribution of Retail Transaction Amounts")
plt.xlabel("Total Spent")
plt.ylabel("Number of Transactions")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

bin_sizes = [5, 10, 20]

for ax, bins in zip(axes, bin_sizes):
    ax.hist(df["Total Spent"], bins=bins, edgecolor="black")
    ax.set_title(f"Histogram with {bins} Bins")
    ax.set_xlabel("Total Spent")
    ax.set_ylabel("Transactions")

plt.tight_layout()
plt.show()
```

```python

mean_spent = df["Total Spent"].mean()
median_spent = df["Total Spent"].median()

plt.figure(figsize=(10, 5))

plt.hist(df["Total Spent"], bins=20, edgecolor="black", alpha=0.7)

plt.axvline(
    mean_spent,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_spent:.2f}"
)

plt.axvline(
    median_spent,
    color="green",
    linestyle="-.",
    linewidth=2,
    label=f"Median = {median_spent:.2f}"
)

plt.title("Transaction Distribution: Mean vs Median")
plt.xlabel("Total Spent")
plt.ylabel("Number of Transactions")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

print("Mean:", round(mean_spent, 2))
print("Median:", round(median_spent, 2))
```

```python

plt.figure(figsize=(10, 5))

plt.hist(
    df["Total Spent"],
    bins=[0, 50, 100, 150, 200, 250, 300, 350, 400, 450],
    edgecolor="black",
    color="skyblue"
)

plt.title("Retail Transactions by Spending Range")
plt.xlabel("Transaction Amount")
plt.ylabel("Number of Transactions")
plt.xticks(range(0, 451, 50))
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python

spending_bins = [0, 50, 100, 150, 200, 250, 300, 350, 400, 450]

spending_ranges = pd.cut(
    df["Total Spent"],
    bins=spending_bins,
    right=True
)

frequency = spending_ranges.value_counts(sort=False)

print("Transactions in Each Spending Range:\n")
print(frequency)

print("\nTotal transactions counted:", frequency.sum())
print("Dataset transactions:", len(df))
```

```python

plt.figure(figsize=(10, 5))

frequency.plot(kind="bar", color="cornflowerblue", edgecolor="black")

plt.title("Number of Transactions by Spending Range")
plt.xlabel("Spending Range")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.hist(
    df["Total Spent"],
    bins=[0, 50, 100, 150, 200, 250, 300, 350, 400, 450],
    cumulative=True,
    edgecolor="black",
    color="mediumseagreen"
)

plt.title("Cumulative Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Cumulative Number of Transactions")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
