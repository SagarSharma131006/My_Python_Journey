```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_store_sales_cleaned.csv")

sns.set_theme(style="whitegrid")

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Total Spent",
    bins=30,
    kde=True,
    color="teal"
)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Total Spent")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df,
    x="Category",
    y="Total Spent",
    inner="quartile",
    color="skyblue",
    cut=0
)

plt.title("Transaction Amount Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Total Spent")
plt.xticks(rotation=35)

plt.tight_layout()
plt.show()
```

```python
numeric_df = df[
    ["Quantity", "Price Per Unit", "Total Spent"]
].dropna()

sns.pairplot(
    numeric_df,
    diag_kind="hist",
    corner=True,
    plot_kws={"alpha": 0.4, "s": 20}
)

plt.suptitle(
    "Relationships Between Numerical Variables",
    y=1.02
)

plt.show()
```

```python
sns.jointplot(
    data=df,
    x="Quantity",
    y="Total Spent",
    kind="reg",
    height=7,
    scatter_kws={"alpha": 0.4, "s": 20},
    line_kws={"color": "red"}
)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.kdeplot(
    data=df,
    x="Total Spent",
    hue="Category",
    fill=False,
    common_norm=False
)

plt.title("Transaction Spending Distribution by Category")
plt.xlabel("Total Spent")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```
