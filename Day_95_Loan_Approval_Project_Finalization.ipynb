```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score, classification_report

from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
# Separate features and target
X = df.drop(columns=["Loan_ID", "Loan_Status"])
y = df["Loan_Status"]

# Identify column types
numerical_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
print("Features prepared successfully!")
```

```python
# Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numerical_cols),
    ("cat", categorical_pipeline, categorical_cols)
])

# Create final model pipeline
final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

print("Final Logistic Regression pipeline created successfully!")
```

```python
# Train the final model
final_model.fit(X_train, y_train)

# Make predictions on test data
y_pred = final_model.predict(X_test)

# Evaluate the model
final_accuracy = accuracy_score(y_test, y_pred)

print("Final Model Trained Successfully!")
print(f"Final Test Accuracy: {final_accuracy:.4f}")
print(f"Final Test Accuracy (%): {final_accuracy * 100:.2f}%")
```

```python
# Save the complete trained pipeline
model_filename = "loan_approval_model.pkl"

joblib.dump(final_model, model_filename)

print(f"Model saved successfully as: {model_filename}")
```

```python
# Load the saved model
loaded_model = joblib.load("loan_approval_model.pkl")

# Predict using the loaded model
loaded_predictions = loaded_model.predict(X_test)

# Compare original and loaded model predictions
predictions_match = np.array_equal(y_pred, loaded_predictions)

print("Model loaded successfully!")
print("Predictions match:", predictions_match)
print("Loaded Model Accuracy:", accuracy_score(y_test, loaded_predictions))
```
