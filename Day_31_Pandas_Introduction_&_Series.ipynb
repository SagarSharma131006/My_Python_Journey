#5️⃣ Understanding Series

```python
marks = pd.Series([78, 85, 92, 67, 74])

print("Values:")
print(marks.values)

print("\nIndex:")
print(marks.index)
```

#🧠 Important

##Series internally:

###Index → Value
####  0   → 78

#6️⃣ Accessing Series Elements

##Normal indexing:

```python
marks = pd.Series([78, 85, 92, 67, 74])

print(marks[0])
print(marks[2])
print(marks[4])
```

#7️⃣ Negative Indexing

##Python lists me:
arr[-1]

##last element deta hai.

#But Pandas Series me .iloc use karna safer and clearer hai:

```python
marks = pd.Series([78, 85, 92, 67, 74])

print(marks.iloc[-1])
print(marks.iloc[-2])
```

#🧠 Important

##iloc = position based indexing

###marks.iloc[0]
###marks.iloc[2]
###marks.iloc[-1]

#8️⃣ Custom Index

##Hum apne indexes bhi define kar sakte hain.

```python
marks = pd.Series(
    [78, 85, 92, 67, 74],
    index=["Maths", "Python", "DBMS", "CN", "OS"]
)

print(marks)
```

##Ab hum label se value access kar sakte hain:

```python
print(marks["Python"])
print(marks["DBMS"])
```

#9️⃣ .loc — Label Based Indexing

##Custom labels ke saath .loc use karna important hai.

```python
marks = pd.Series(
    [78, 85, 92, 67, 74],
    index=["Maths", "Python", "DBMS", "CN", "OS"]
)

print(marks.loc["Python"])
print(marks.loc["DBMS"])
```

#🧠 Remember
##iloc → position
##loc  → label

#🔟 Series Slicing

```python
marks = pd.Series([78, 85, 92, 67, 74])

print(marks[1:4])
```

##Using .iloc:

```python
print(marks.iloc[1:4])
```

#1️⃣1️⃣ Series Properties

##Pandas Series ki important properties:

```python
marks = pd.Series([78, 85, 92, 67, 74])

print("Shape:", marks.shape)
print("Size:", marks.size)
print("Dimension:", marks.ndim)
print("Data Type:", marks.dtype)
```

#1️⃣2️⃣ Series Mathematical Operations

```python
marks = pd.Series([50, 60, 70, 80, 90])

print(marks + 5)
print(marks * 2)
print(marks / 10)
```

#🧠 Important

##Pandas Series bhi vectorized operations support karti hai.

#1️⃣3️⃣ Series Statistical Functions

```python
marks = pd.Series([78, 85, 92, 67, 74])

print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Maximum:", marks.max())
print("Minimum:", marks.min())
print("Standard Deviation:", marks.std())
```

#🧠 Mind Template
###sum()  → total
###mean() → average
###max()  → highest
###min()  → lowest
###std()  → variation

#1️⃣4️⃣ Boolean Filtering

##Ab Pandas ka important feature.

```python
marks = pd.Series([45, 67, 89, 72, 95, 54, 81])

high_marks = marks[marks > 75]

print(high_marks)
```

#Notice karo:

##value ke saath original index bhi preserve hua.

#1️⃣5️⃣ Multiple Conditions

##Suppose marks between 60 and 90 chahiye.

```python
marks = pd.Series([45, 67, 89, 72, 95, 54, 81])
result = marks[(marks >= 60) & (marks <= 90)]
print(result)
```

#🧠 Important

##Pandas me bhi:

###&
###|
###~

###use karte hain

#1️⃣8️⃣ Practical Example — Student Marks

##Ab thoda real-data style.

```python
import pandas as pd

marks = pd.Series(
    [78, 85, 92, 67, 74],
    index=["Sagar", "Rahul", "Aman", "Priya", "Neha"]
)

print("All Marks:")
print(marks)

print("\nAverage Marks:")
print(marks.mean())

print("\nHighest Marks:")
print(marks.max())

print("\nStudents above 80:")
print(marks[marks > 80])
```

#Q10 — 🔥 Mini Challenge

##Create a Student Marks Analyzer using Pandas Series.

##Your program should:

##Create student marks
###       ↓
##Display marks
###       ↓
##Calculate average
###       ↓
##Find highest marks
###       ↓
##Find lowest marks
###       ↓
##Filter students above 80
###       ↓
##Show statistical summary

```python
import pandas as pd

marks = pd.Series(
    [85, 72, 91, 64, 78],
    index=["Rahul", "Aman", "Priya", "Neha", "Rohit"]
)

print("Student Marks:")
print(marks)

print("\nAverage Marks:", marks.mean())
print("Highest Marks:", marks.max())
print("Lowest Marks:", marks.min())

print("\nStudents scoring above 75:")
print(marks[marks > 75])

print("\nStudents scoring 80 or above:")
print(marks[marks >= 80])
```
