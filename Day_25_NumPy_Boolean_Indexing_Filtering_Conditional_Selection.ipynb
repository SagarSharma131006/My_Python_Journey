# 1. Boolean Conditions in NumPy

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr > 30)
```

#🧠 Important Point

###NumPy comparison operation ka result ek Boolean array hota hai.

# 2. Different Comparison Operators

###NumPy mein normal comparison operators use kar sakte hain:

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr > 30)
print(arr < 30)
print(arr >= 30)
print(arr <= 30)
print(arr == 30)
print(arr != 30)
```

# 3. Boolean Array

```python
arr = np.array([10, 20, 30, 40, 50])

condition = arr > 30

print(condition)
```

# 4. Boolean Indexing

###Ab sabse important part.

###Hum Boolean array ko original array ke andar use kar sakte hain:

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[arr > 30])
```

# 5. Filter Values Greater Than a Number

```python
arr = np.array([12, 45, 23, 67, 34, 89, 10])

result = arr[arr > 40]

print(result)
```

# 6. Filter Values Less Than a Number

```python
arr = np.array([12, 45, 23, 67, 34, 89, 10])

print(arr[arr < 40])
```

# 7. Filter Exact Value

###Suppose hume sirf 30 find karna hai:

```python
arr = np.array([10, 20, 30, 40, 30, 50])

print(arr[arr == 30])
```

# 8. Filtering with Range

##Suppose hume values chahiye:

###20 se greater
###AND
###50 se less

```python
arr = np.array([10, 20, 25, 30, 45, 50, 60])

result = arr[(arr > 20) & (arr < 50)]

print(result)
```

# 9. AND Condition — &

```python
arr = np.array([10, 20, 30, 40, 50, 60])

result = arr[(arr >= 20) & (arr <= 50)]

print(result)
```

# 10. OR Condition — |

##Suppose hume values chahiye jo:

###20 se less
###OR
###50 se greater

```python
arr = np.array([10, 20, 30, 40, 50, 60])

result = arr[(arr < 20) | (arr > 50)]

print(result)
```

# 11. NOT Condition — ~

###~ ka use condition ko reverse karne ke liye hota hai.

```python
arr = np.array([10, 20, 30, 40, 50])

condition = arr > 30

print(condition)
print(~condition)
```

# 12. Why Parentheses Are Important

##Ye sahi hai:

arr[(arr > 20) & (arr < 50)]

##Ye mat likhna:

arr[arr > 20 & arr < 50]

###Multiple NumPy conditions mein each condition ko parentheses mein rakhna safest and standard approach hai.

# 13. np.where()

###np.where() condition ke basis par values/indices determine karne ke liye useful hai.

```python
arr = np.array([10, 20, 30, 40, 50])

result = np.where(arr > 30)

print(result)
```

# 14. np.where() with Two Values

###np.where() ka ek aur powerful use:

```python
arr = np.array([10, 20, 30, 40, 50])

result = np.where(arr > 30, 1, 0)

print(result)
```

#Q10 — Mini Practical

##Temperature data:

temperature = np.array([
    18, 22, 25, 31, 35, 28, 40, 19, 33, 27
])

##Find:

###Temperatures above 30
###Temperatures below 20
###Temperatures between 20 and 30, both included
###Temperatures above 25 and below 35

```python
temperature = np.array([ 18, 22, 25, 31, 35, 28, 40, 19, 33, 27 ])
print("Temperatures above 30:",temperature[temperature>30])
print("Temperatures below 20:",temperature[temperature<20])
print("Temperatures between 20 and 30, both included:",temperature[(temperature>=20)&(temperature<=30)])
print("Temperatures above 25 and below 35:",temperature[(temperature>25)&(temperature<35)])
```
