# 1. NumPy Sorting

`np.sort()` ka use array ke elements ko ascending order mein arrange karne ke liye hota hai.

### Syntax

```python
np.sort(array)
```

````python
import numpy as np

arr = np.array([50, 20, 40, 10, 30])

print(np.sort(arr))
````

# 🧠 Important Point

`np.sort(arr)` sorted array return karta hai aur original `arr` ko change nahi karta.

# 2. Sorting in Descending Order

NumPy mein descending sort ke liye pehle ascending sort karo aur phir slicing se reverse karo.

````python
arr = np.array([50, 20, 40, 10, 30])

sorted_arr = np.sort(arr)

print(sorted_arr[::-1])
````

# 3. Sorting a 2D Array

2D array ko sort karte waqt `axis` decide karta hai ki sorting rows ke along hogi ya columns ke along.

### `axis=1`

Har row ke elements independently sort honge.

````python
arr = np.array([
    [30, 10, 20],
    [60, 40, 50],
    [90, 70, 80]
])

print(np.sort(arr, axis=1))
````

# 🧠 Important Point

```text
axis=1 → each row
axis=0 → each column
```

# 4. Sorting 2D Array by Columns

`axis=0` ka use karke har column ko independently sort kar sakte hain.

````python
arr = np.array([
    [30, 10, 20],
    [60, 40, 50],
    [90, 70, 80]
])

print(np.sort(arr, axis=0))
````

# 5. `np.argsort()`

`np.argsort()` sorted values nahi, balki **sorted order mein indices** return karta hai.

````python
arr = np.array([40, 10, 30, 20])

indices = np.argsort(arr)

print(indices)
print(arr[indices])
````

# 🧠 Important Point

```text
np.sort()    → sorted values
np.argsort() → sorted positions/indices
```

# 6. `np.argmax()` and `np.argmin()`

`np.argmax()` maximum value ka index return karta hai.

`np.argmin()` minimum value ka index return karta hai.

````python
arr = np.array([45, 12, 89, 34, 67])

print("Maximum value:", arr.max())
print("Maximum index:", np.argmax(arr))

print("Minimum value:", arr.min())
print("Minimum index:", np.argmin(arr))
````

# 7. Finding Positions with `np.where()`

`np.where()` ka use condition ke basis par matching indices find karne ke liye bhi kiya ja sakta hai.

````python
arr = np.array([10, 20, 30, 20, 40, 20])

positions = np.where(arr == 20)

print(positions)
````

# 8. `np.searchsorted()`

`np.searchsorted()` sorted array mein kisi value ko insert karne ke liye suitable position batata hai.

Array sorted hona chahiye.

````python
arr = np.array([10, 20, 30, 40, 50])

print(np.searchsorted(arr, 35))
print(np.searchsorted(arr, 25))
````

# 9. `np.unique()`

`np.unique()` array ke duplicate values ko remove karke unique values return karta hai.

````python
arr = np.array([10, 20, 20, 30, 10, 40, 30])

print(np.unique(arr))
````

# 10. Unique Values with Counts

`return_counts=True` ke saath har unique value ki frequency bhi mil sakti hai.

````python
arr = np.array([10, 20, 20, 30, 10, 40, 30])

values, counts = np.unique(arr, return_counts=True)

print("Values:", values)
print("Counts:", counts)
````

# 11. Practical Example — Student Marks

````python
marks = np.array([78, 92, 65, 88, 92, 74, 65, 95])

print("Sorted Marks:", np.sort(marks))
print("Highest Marks:", marks.max())
print("Highest Marks Index:", np.argmax(marks))
print("Unique Marks:", np.unique(marks))

values, counts = np.unique(marks, return_counts=True)

print("Values:", values)
print("Counts:", counts)
````

# 12. Practical Example — Product Prices

Suppose ek shop mein products ki prices stored hain.

````python
prices = np.array([499, 999, 299, 799, 499, 1299, 299])

print("Sorted Prices:", np.sort(prices))
print("Lowest Price:", prices.min())
print("Highest Price:", prices.max())
print("Unique Prices:", np.unique(prices))
````
