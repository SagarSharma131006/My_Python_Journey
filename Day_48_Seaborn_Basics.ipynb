```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_store_sales_cleaned.csv")

print("Dataset shape:", df.shape)
print("Seaborn version:", sns.__version__)
```

```python
plt.figure(figsize=(11, 6))

sns.barplot(
    data=df,
    x="Category",
    y="Total Spent",
    estimator="sum",
    errorbar=None,
    color="steelblue"
)

plt.title("Total Sales by Category", fontsize=15, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=35, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11, 6))

sns.countplot(
    data=df,
    x="Category",
    order=df["Category"].value_counts().index,
    color="teal"
)

plt.title("Number of Transactions by Category",
          fontsize=15, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Transaction Count")
plt.xticks(rotation=35, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="Category",
    y="Total Spent",
    color="lightblue"
)

plt.title("Transaction Spending by Category",
          fontsize=15, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Total Spent")
plt.xticks(rotation=35, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Total Spent",
    hue="Category",
    alpha=0.6,
    s=50
)

plt.title("Quantity vs Total Spent by Category",
          fontsize=15, fontweight="bold")
plt.xlabel("Quantity")
plt.ylabel("Total Spent")
plt.legend(title="Category", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
numeric_cols = ["Quantity", "Price Per Unit", "Total Spent"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".3f"
)

plt.title("Correlation Heatmap", fontsize=15, fontweight="bold")
plt.tight_layout()
plt.show()
```
