```python
!pip -q install streamlit

app_code = '''
import streamlit as st

st.set_page_config(page_title="My First ML App", page_icon="🤖")

st.title("🤖 My First Machine Learning App")
st.write("Welcome to my first Streamlit application!")
st.write("I am learning Machine Learning deployment.")

name = st.text_input("Enter your name")
age = st.slider("Select your age", 18, 60, 22)

course = st.selectbox(
    "Choose your field",
    ["Data Science", "Machine Learning", "AI Engineering"]
)

if st.button("Submit"):
    st.success(f"Hello {name}!")
    st.write("Your Age:", age)
    st.write("Your Selected Field:", course)
    st.info("Streamlit app is working successfully! 🚀")
'''

with open("app.py", "w") as file:
    file.write(app_code)

print("Streamlit installed successfully!")
print("app.py created successfully! ✅")
```

```python
import subprocess
import time
import urllib.request
import os

# Check if Streamlit is already working
try:
    urllib.request.urlopen("http://localhost:8501", timeout=3)
    print("Streamlit is already running ✅")
except:
    print("Starting Streamlit server...")
    subprocess.Popen(
        ["streamlit", "run", "app.py", "--server.port", "8501", "--server.headless", "true"],
        stdout=open("streamlit.log", "w"),
        stderr=subprocess.STDOUT
    )
    time.sleep(10)

# Start Cloudflare tunnel
print("Starting Cloudflare tunnel...")

tunnel = subprocess.Popen(
    ["./cloudflared", "tunnel", "--url", "http://127.0.0.1:8501"],
    stdout=open("tunnel.log", "w"),
    stderr=subprocess.STDOUT
)

# Wait for public URL
url = None

for _ in range(30):
    time.sleep(2)
    if os.path.exists("tunnel.log"):
        with open("tunnel.log", "r") as f:
            log = f.read()
        for line in log.splitlines():
            if "trycloudflare.com" in line:
                import re
                match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", line)
                if match:
                    url = match.group(0)
                    break
    if url:
        break

if url:
    print("\n🚀 YOUR STREAMLIT APP IS LIVE:")
    print(url)
else:
    print("Tunnel URL not found. Check tunnel.log for errors.")
```
