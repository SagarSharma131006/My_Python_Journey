#Missing Values Analysis 🔍

```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:\n")
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

#Missing Values Percentage

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": missing_percent
})

missing_summary = missing_summary[missing_summary["Missing_Count"] > 0]

missing_summary.sort_values(by="Missing_Percentage", ascending=False)
```

#Missing Values ko visualize karna

```python
plt.figure(figsize=(12, 5))

sns.heatmap(df.isnull(), cbar=False, cmap="viridis")

plt.title("Missing Values in Loan Dataset")
plt.xlabel("Columns")
plt.show()
```

#Target Variable Analysis

##1. Count plot

```python
plt.figure(figsize=(6, 4))

sns.countplot(x="Loan_Status", data=df)

plt.title("Loan Approval Status Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Number of Applicants")
plt.show()
```

#Exact counts & percentage

```python
print("Loan Status Counts:\n")
print(df["Loan_Status"].value_counts())

print("\nLoan Status Percentage:\n")
print((df["Loan_Status"].value_counts(normalize=True) * 100).round(2))
```
