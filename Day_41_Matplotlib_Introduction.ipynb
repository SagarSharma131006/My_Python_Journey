#1. Line Plot :

```python
# Sample data
days = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 200, 250]

# Create line graph
plt.plot(days, sales)

# Display graph
plt.show()
```

##Title and Axis Label :

```python
days = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 200, 250]

plt.plot(days, sales)

plt.title("Daily Sales Trend")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
```

##Markers , Line Style and Grid

```python
days = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 200, 250]

plt.plot(days, sales, marker='o', linestyle='-', color='green')

plt.title("Daily Sales Trend")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

##Figure Size and Line Width

```python

days = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 200, 250]

plt.figure(figsize=(8, 4))

plt.plot(
    days,
    sales,
    marker='o',
    linestyle='-',
    color='green',
    linewidth=2
)

plt.title("Daily Sales Trend")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

##Multiple Lines and Legend

```python

days = [1, 2, 3, 4, 5]
store_a = [100, 150, 120, 200, 250]
store_b = [80, 130, 160, 180, 220]

plt.figure(figsize=(8, 4))

plt.plot(days, store_a, marker='o', label='Store A')
plt.plot(days, store_b, marker='s', label='Store B')

plt.title("Daily Sales Comparison")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)
plt.show()
```

##Different Color and Line Style

```python

days = [1, 2, 3, 4, 5]
store_a = [100, 150, 120, 200, 250]
store_b = [80, 130, 160, 180, 220]

plt.figure(figsize=(8, 4))

plt.plot(
    days, store_a,
    color='blue',
    linestyle='-',
    linewidth=2,
    marker='o',
    label='Store A'
)

plt.plot(
    days, store_b,
    color='red',
    linestyle='--',
    linewidth=2,
    marker='s',
    label='Store B'
)

plt.title("Daily Sales Comparison")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, linestyle=':')
plt.show()
```

```python

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales_2025 = [120, 150, 135, 180, 220]
sales_2026 = [140, 160, 170, 200, 250]

# 1. Set figure size
plt.figure(figsize=(9, 5))

# 2. Plot sales for 2025
plt.plot(
    months, sales_2025,
    color='green',
    linestyle='-',
    marker='o',
    label='2025'
)

# 3. Plot sales for 2026
plt.plot(
    months, sales_2026,
    color='blue',
    linestyle='--',
    marker='s',
    label='2026'
)

# 4. Add title
plt.title("Monthly Sales Comparison")

# 5. X-axis label
plt.xlabel("Month")

# 6. Y-axis label
plt.ylabel("Sales")

# 7. Show legend and grid
plt.legend()
plt.grid(True)

# 8. Display graph
plt.show()
```
