#1️⃣2️⃣ Conditional Selection

##Ab real Data Analysis wali power start hoti hai.

##Suppose humein students chahiye jinke Python marks 80 se greater hain.

```python
print(students[students["Python"] > 80])
```

#🧠 Mind Template:

Condition

↓

students["Python"] > 80

↓

Boolean Series

↓

students[condition]

↓

Filtered DataFrame

#1️⃣3️⃣ Python Marks >= 80

```python
high_python = students[students["Python"] >= 80]

print(high_python)
```

#1️⃣4️⃣ Python Marks < 70

```python
print(students[students["Python"] < 70])
```

#1️⃣5️⃣ Multiple Conditions

Suppose:

Python >= 80 AND Math >= 80

```python
result = students[
    (students["Python"] >= 80) &
    (students["Math"] >= 80)
]

print(result)
```

#1️⃣6️⃣ OR Condition

##Python >= 90 OR Math >= 90:

```python
result = students[
    (students["Python"] >= 90) |
    (students["Math"] >= 90)
]

print(result)
```

#1️⃣7️⃣ NOT Condition

##Python marks 80 or above ko exclude karna:

```python
result = students[
    ~(students["Python"] >= 80)
]

print(result)
```

#1️⃣8️⃣ isin()

##Suppose humein sirf Rahul aur Priya ka data chahiye.

```python
result = students[
    students["Name"].isin(["Rahul", "Priya"])
]

print(result)
```

#1️⃣9️⃣ query()

##Pandas mein conditions likhne ka ek clean way:

```python
result = students.query("Python >= 80")

print(result)
```

```python
result = students.query(
    "Python >= 80 and Math >= 80"
)

print(result)
```

#2️⃣0️⃣ Selecting Specific Columns After Filtering

##Suppose humein sirf Name + Python chahiye jahan Python >= 80 hai:

```python
result = students.loc[
    students["Python"] >= 80,
    ["Name", "Python"]
]

print(result)
```

#2️⃣1️⃣ Practical Data Analysis Example

##Suppose company ke employees ka dataset hai:

```python
employees = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha", "Rohit"],
    "Department": ["IT", "HR", "IT", "Finance", "IT"],
    "Salary": [50000, 45000, 70000, 55000, 65000],
    "Experience": [2, 3, 5, 4, 6]
})

print(employees)
```

```python
# IT EMPLOYEE
it_employees = employees[
    employees["Department"] == "IT"
]

print(it_employees)
```

```python
# SALARY>60000
high_salary = employees[
    employees["Salary"] > 60000
]

print(high_salary)
```

```python
# IT + SALARY>60000
result = employees[
    (employees["Department"] == "IT") &
    (employees["Salary"] > 60000)
]

print(result)
```
