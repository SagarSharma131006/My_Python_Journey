## 1. Train-Test Split & Generalization

- **Training set:** the model learns patterns from this data.
- **Test set:** held-out data used to estimate performance on unseen examples.
- **Generalization:** how well the trained model performs on new, unseen data.

If we test on the same data used for training, the evaluation can be overly optimistic and may not reveal whether the model generalizes.

## 2. StandardScaler

StandardScaler transforms features so they are centered around 0 and scaled to unit standard deviation (when possible). This helps prevent features with larger numeric scales from dominating certain algorithms.

Scaling is particularly important for models/methods such as **Logistic Regression, SVM, KNN, and PCA**. Tree-based models usually don't require standardization.

Conceptually: `z = (x - mean) / standard_deviation`

## 3. Evaluation Metrics

### Accuracy
Fraction of all predictions that are correct. Accuracy alone can be misleading when classes are imbalanced.

### Confusion-matrix terms
- **TP (True Positive):** correctly predicted positive.
- **FP (False Positive):** negative example incorrectly predicted positive (false alarm).
- **TN (True Negative):** correctly predicted negative.
- **FN (False Negative):** positive example incorrectly predicted negative (missed case).

### Precision vs Recall
- **Precision:** Among predictions marked positive, how many are actually positive?
- **Recall:** Among all actual positive cases, how many did the model catch?


`Precision = TP / (TP + FP)`

`Recall = TP / (TP + FN)`

```python
# Example: 60 true positives, 20 false positives, 40 false negatives
TP, FP, FN = 60, 20, 40
precision = TP / (TP + FP)
recall = TP / (TP + FN)
print(f"Precision: {precision:.0%}")
print(f"Recall: {recall:.0%}")
```

## 4. Choosing a metric for the situation

- **Fraud detection (catch as many fraud cases as possible):** prioritize Recall, because missing actual fraud cases (FN) is costly.
- **Spam detection (avoid sending legitimate emails to spam):** prioritize Precision for the spam-positive class, so emails marked spam are more likely to truly be spam.

Metric choice depends on which kind of mistake matters most.

## 5. Overfitting vs Underfitting

| Condition | Training performance | Test performance |
|---|---|---|
| Overfitting | Very high / excellent | Much lower / poor |
| Underfitting | Poor | Poor |
| Good fit | Good | Similar, good performance |

**Overfitting:** model learns training examples too specifically and does not generalize well.

**Underfitting:** model fails to capture important patterns, even in training data.

## 6. Ways to reduce overfitting

1. Reduce model complexity.
2. Collect more representative training data.
3. Use regularization.
4. Use cross-validation for model selection.
5. Remove noisy or unnecessary features when justified.

These are options to investigate; the best choice depends on the model and data.

## 7. Day 90 quiz — questions and your final answers

### Q1. Why use `train_test_split`?
**Answer:** A separate test set helps estimate how well the model will generalize to unseen data. Testing on training data can give an overly optimistic result.

### Q2. What does `StandardScaler` do?
**Answer:** Standardizes feature scales so large-valued features don't dominate certain models just because of their units or magnitude. Particularly useful for Logistic Regression, SVM, KNN, and PCA.

### Q3. Is 98% accuracy always excellent?
**Answer:** No. With imbalanced classes, a model can have high accuracy while missing the minority class. Check the confusion matrix and other relevant metrics (Precision, Recall, F1, etc.).

### Q4. Overfitting vs Underfitting
**Answer:** Overfitting means excellent training performance but poorer test performance. Underfitting means poor performance on both training and test sets.

### Q5. Precision vs Recall
**Answer:** Precision asks “among predicted positives, how many were correct?” Recall asks “among actual positives, how many were caught?”
