# 1. NumPy kya hai?

NumPy = Numerical Python

Ye Python ki ek library hai jo mainly numbers, arrays aur numerical calculations ke liye use hoti hai.

```python
# Normal Python List:
numbers=[10,20,30,40,50]

# NumPy Array:
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])
```

NumPy arrays ka sabse important concept hai:

NumPy ek powerful multidimensional array structure provide karta hai.

Ye Data Science, Machine Learning, scientific computing etc. mein bahut commonly use hota hai.

# 🔥 Ek saath sab properties

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:")
print(arr)

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data Type:", arr.dtype)
```
